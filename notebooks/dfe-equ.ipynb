# XCZU67DR complex equalizer
One antenna, four complex taps and a 100 MHz sample clock. The DMA sends known IQ samples. Each capture starts at the playback boundary. The identity impulse supplies a single pipeline delay used in every comparison. The intentional delay in the second filter stays in its frequency response.

Before acquisition the acceptance limits are ±0.1 dB and ±1° at each of five signed frequencies and ≤2 LSB per I or Q component on impulse and step responses. Frequencies are coherent with the observation window. The tests use coefficient shift 6 and scale integer coefficients by 1/16384.

```python
import os, sys
from pathlib import Path
examples = Path('../packages/pynq-dfe/examples/equ')
if examples.exists():
    sys.path.insert(0, str(examples.resolve()))
from equalizer import measure
out = Path(os.getenv('EQU_OUTPUT', '.'))
report = measure(Path(os.getenv('EQU_BITFILE', 'dfe_equ.bit')), out)
print('Software and core versions', report['versions'])
print('Latency (clocks), register and impulse', report['hardware_latency'], report['measured_latency'])
```

```python
print('filter       MHz     pred dB  meas dB  pred deg  meas deg  result')
for row in report['rows']:
    verdict = 'PASS' if row['passed'] else 'FAIL'
    if 'MHz' in row:
        print('{filter:10} {MHz:7.3f} {predicted_dB:8.3f} {measured_dB:8.3f} {predicted_deg:9.3f} {measured_deg:9.3f}'.format(**row), verdict)
    else:
        print(row['filter'], row['case'], row['max_error_lsb'], 'LSB', verdict)
assert all(row['passed'] for row in report['rows'])
```
